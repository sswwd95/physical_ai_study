```python
import cv2
```

```python
# 컬러 이미지 읽기
img_color = cv2.imread(r"C:\work\physical_ai\opencv\practice_images\sample.jpg", cv2.IMREAD_COLOR)
cv2.namedWindow("Color Image")
cv2.imshow("Color Image", img_color)
cv2.waitKey(0) # 실시간 카메라에서는 waitKey(0)이 아니라 보통 waitKey(1)을 사용
cv2.destroyAllWindows()
```

```python
# 이미지 흑백으로 변환
img_gray = cv2.imread(r"C:\work\physical_ai\opencv\practice_images\sample.jpg", cv2.IMREAD_GRAYSCALE )
cv2.namedWindow("Color Image")
cv2.imshow("Color Image", img_gray)
cv2.waitKey(0) # 실시간 카메라에서는 waitKey(0)이 아니라 보통 waitKey(1)을 사용
cv2.destroyAllWindows()
```

```python
cap = cv2.VideoCapture(0) # 0번 카메라 장치 열기
if cap.isOpened() == False:
    print("카메라를 열 수 없습니다.")
    exit(1)
while (True):
    ret, frame = cap.read() # 카메라에서 프레임 읽기
    if ret == False:
        print("카메라에서 프레임을 읽을 수 없습니다.")
        break
    cv2.imshow("Camera", frame) # 프레임 출력
    if cv2.waitKey(1) == 27: # ESC 키를 누르면 종료
        break

cap.release()
cv2.destroyAllWindows()
```
