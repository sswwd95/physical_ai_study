# 8. 로그에 센서 데이터를 모두 남길 때의 문제

100Hz 센서를 8시간 기록한다고 가정합니다.

```
100건/초 × 60초 × 60분 × 8시간
= 2,880,000건
```

관절 7개, 토크, 속도, 위치를 모두 기록하면 로그가 매우 커집니다.

따라서 다음 전략을 사용합니다.

```
원본 센서 데이터
→ CSV, Parquet, 데이터베이스 저장

일반 로그
→ 시작, 종료, 경고, 오류만 저장

지표 로그
→ 평균, 최대값, RMS, 이상 점수 저장
```
예:

```python
logger.debug(
    "센서 배치 처리 완료 | "
    "rows=%d | mean=%.3f | max=%.3f",
    len(values),
    mean_value,
    maximum_value,
)
```

센서값 10,000개를 일반 로그에 모두 출력하지 않습니다.

# 9. 실무에서 자주 하는 로깅 실수

## 실수 1. 모든 메시지를 INFO로 기록

잘못된 예:

```python
logger.info("충돌 발생")
logger.info("시스템 안전 정지 실패")
```

개선:

```python
logger.error("충돌 발생")
logger.critical("시스템 안전 정지 실패")
```

## 실수 2. 오류 메시지만 기록하고 스택을 누락
잘못된 예:

```python
logger.error(str(error))
```

개선:

```python
logger.exception("Panda 제어 실패")
```

## 실수 3. 실행 ID가 없음

여러 실험의 로그가 섞여 결과 비교가 어려워집니다.

개선:

```
run_id
experiment_name
robot_id
task_name
```
## 실수 4. 로그 파일 무제한 증가

개선:

```python
RotatingFileHandler
```

또는 날짜 기반 회전이 필요하면 다음 클래스를 사용할 수 있습니다.

```python
TimedRotatingFileHandler
```

## 실수 5. 센서 원본값을 지나치게 출력

개선:

```
원본 데이터는 별도 데이터 파일
로그에는 요약값과 이상 이벤트
```
## 실수 6. 로그 파일과 지표 파일을 구분하지 않음

개선:

```
logs/
metrics/
reports/
```

## 실수 7. 작업 성공 로그만 있고 실패 상태가 없음

실행 종료 시 반드시 성공 여부를 남깁니다.

````python
from pathlib import Path
from random import Random
from time import perf_counter, sleep

from src.harness.logging_harness import (
    ManufacturingLoggingHarness,
)


def run_multiple_cycles() -> None:
    harness = ManufacturingLoggingHarness(
        runs_root=Path("runs"),
        project_name="manufacturing_physical_ai",
        experiment_name="panda_five_cycles",
    )

    logger = harness.get_task_logger(
        task_name="production_loop",
        robot_id="panda_01",
    )

    random_generator = Random(42)

    try:
        defect_count = 0

        for cycle_id in range(1, 6):
            start_time = perf_counter()

            temperature = random_generator.uniform(
                55.0,
                78.0,
            )

            vibration = random_generator.uniform(
                0.10,
                0.30,
            )

            motor_current = random_generator.uniform(
                4.5,
                8.5,
            )

            defect = (
                random_generator.random() < 0.1
            )

            sleep(0.05)

            cycle_time = (
                perf_counter() - start_time
            )

            if temperature >= 75.0:
                logger.error(
                    "온도 위험 | cycle=%d | "
                    "temperature=%.2f",
                    cycle_id,
                    temperature,
                )

            elif temperature >= 65.0:
                logger.warning(
                    "온도 경고 | cycle=%d | "
                    "temperature=%.2f",
                    cycle_id,
                    temperature,
                )

            else:
                logger.info(
                    "온도 정상 | cycle=%d | "
                    "temperature=%.2f",
                    cycle_id,
                    temperature,
                )

            if defect:
                defect_count += 1

                harness.log_event(
                    message="불량 발생",
                    event_type="defect_detected",
                    fields={
                        "cycle_id": cycle_id,
                        "robot_id": "panda_01",
                    },
                )

            metrics = {
                "production.cycle_time_seconds": (
                    cycle_time
                ),
                "sensor.temperature": temperature,
                "sensor.vibration": vibration,
                "sensor.motor_current": (
                    motor_current
                ),
                "quality.defect": int(defect),
            }

            for metric_name, metric_value in (
                metrics.items()
            ):
                harness.log_metric(
                    metric_name=metric_name,
                    metric_value=metric_value,
                    step=cycle_id,
                    tags={
                        "robot_id": "panda_01",
                        "line_id": "line_01",
                    },
                )

            logger.info(
                "생산 사이클 완료 | cycle=%d | "
                "cycle_time=%.3f | defect=%s",
                cycle_id,
                cycle_time,
                defect,
            )

        defect_rate = defect_count / 5

        harness.log_metric(
            metric_name="quality.defect_rate",
            metric_value=defect_rate,
            step=5,
            tags={
                "line_id": "line_01",
            },
        )

        harness.finish(success=True)

    except Exception as error:
        logger.exception(
            "생산 사이클 반복 실행 실패"
        )

        harness.finish(
            success=False,
            error_message=(
                f"{type(error).__name__}: {error}"
            ),
        )


if __name__ == "__main__":
    run_multiple_cycles()
````
