| --- | --- | --- |
| DEBUG | 개발용 상세 정보 | 관절별 위치값, 센서 원본값 |

| CRITICAL | 시스템 중단 위험 | 안전 정지 실패 |

## 잘못된 방식

```python
try:
    run_simulation()
except Exception as error:
    logger.error(str(error))
```

이 방식은 오류 메시지만 남고, 어느 줄에서 발생했는지 알기 어렵습니다.

## 권장 방식

```python
logger.exception("시뮬레이션 실행 실패")
```

`logger.exception()`은 현재 예외의 스택 추적을 자동으로 기록합니다.

## 예제 소스 생성 프롬프트
```
제조 센서 분석 함수에서 데이터가 비어 있으면 예외를 발생시키고,
logger.exception을 사용하여 전체 스택 추적을 로그 파일에 기록하는
예제를 작성해줘.

오류가 발생해도 프로그램이 종료되지 않고 실패 여부를 반환해야 한다.
```
