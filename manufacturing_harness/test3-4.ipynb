## [실습 23. 로그 레벨별 메시지 관리]

## 핵심 주제

상황에 맞는 로그 레벨을 선택합니다.

## 로그 레벨

| 레벨 | 용도 | 제조 예 |
| --- | --- | --- |
| DEBUG | 개발용 상세 정보 | 관절별 위치값, 센서 원본값 |
| INFO | 정상 처리 정보 | 시뮬레이션 시작, 작업 완료 |
| WARNING | 주의 필요 | 온도 경고 기준 초과 |
| ERROR | 작업 실패 | 센서 파일 읽기 실패 |
| CRITICAL | 시스템 중단 위험 | 안전 정지 실패 |

## 예제 소스 생성 프롬프트
```
제조 생산 라인의 온도값을 입력받아 정상, 경고, 위험 상태를 판단하고
Python logging의 INFO, WARNING, ERROR 레벨로 기록하는 함수를 작성해줘.

조건:
- 정상: warning 미만
- 경고: warning 이상 critical 미만
- 위험: critical 이상
- 입력값과 임계값을 로그에 포함
```
