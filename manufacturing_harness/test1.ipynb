# 1. 설정 관리가 필요한 이유

제조 Physical AI 프로젝트에는 다음과 같은 값이 계속 변경됩니다.

```
MuJoCo 모델 경로
시뮬레이션 시간 간격
제어 주기
센서 샘플링 주기
이상 탐지 임계값
설비 온도 제한
Panda 관절 제한
강화학습 알고리즘
학습 횟수
랜덤 시드
로그 저장 경로
결과 저장 경로
```

이 값을 Python 소스 안에 직접 작성하면 수정할 때마다 코드를 변경해야 합니다.

```python
model_path = "C:/projects/mujoco_menagerie/franka_emika_panda/scene.xml"
temperature_limit = 70.0
control_frequency = 100
```

이를 **하드코딩**이라고 합니다.

설정 파일을 사용하면 소스 코드를 바꾸지 않고 실험 조건만 변경할 수 있습니다.

```yaml
simulation:
  model_path: assets/mujoco_menagerie/franka_emika_panda/scene.xml
  control_frequency: 100

monitoring:
  temperature_limit: 70.0
```
